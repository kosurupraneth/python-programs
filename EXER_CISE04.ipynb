```python
def reverse(s):
    sr =__builtins__.str(s)
    a = []

    for i in range(len(sr) - 1, -1, -1):
        a.append(sr[i])

    print("".join(a))

reverse(345)
```

```python
def reverse(s):
    sr = __builtins__.str(s)
    print(sr[::-1])

reverse(345)
```
