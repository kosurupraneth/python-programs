```python
fruits=["apple","banana","orange"]
with open("f3.txt","w") as f:
  for fruit in fruits:
    f.write(fruit+"\n")
print("file created")
```

```python
with open("f3.txt", "r") as f:
     fruits = f.read().splitlines()

names = [fruit.upper() for fruit in fruits]

print(names)
```
